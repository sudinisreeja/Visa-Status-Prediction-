```python
%%writefile predict.py

import pandas as pd
import joblib

# Load trained model when the script starts
model = joblib.load("rf_model.pkl")

def preprocess_input(data):
    """
    Convert user input into model-ready format, matching the model's expected features.
    """

    # Retrieve the exact feature names the model was trained on
    model_features = model.feature_names_in_

    # Create a new pandas DataFrame with a single row,
    # all columns from model_features, and initialized with zeros.
    # This provides default values for numerical features not supplied by the user
    # and sets all one-hot encoded features to 0 by default.
    processed_df = pd.DataFrame(0, index=[0], columns=model_features)

    # Handle one-hot encoding for 'WORKSITE'
    # The 'country' from input maps to 'WORKSITE' in the model's features
    worksite_col_name = f"WORKSITE_{data['country']}"
    if worksite_col_name in processed_df.columns:
        processed_df[worksite_col_name] = 1

    # Handle one-hot encoding for 'VISA TYPE'
    visa_type_col_name = f"VISA TYPE_{data['visa_type']}"
    if visa_type_col_name in processed_df.columns:
        processed_df[visa_type_col_name] = 1

    # Handle 'application_month' if it is a feature in the model
    if 'application_month' in processed_df.columns:
        application_date_dt = pd.to_datetime(data['application_date'])
        processed_df['application_month'] = application_date_dt.month

    # Other numerical features (like S.NO, AGE, PREVAILING_WAGE, YEAR)
    # and other one-hot encoded categorical features will remain 0 if not explicitly provided
    # in the input 'data' and not handled above. This is the desired default behavior.

    return processed_df

def predict_processing_time(input_data):
    """
    Takes user input -> returns predicted processing time
    """

    processed_data = preprocess_input(input_data)

    prediction = model.predict(processed_data)

    return round(prediction[0], 2)
```
